### `경기도_버스노선 조회` 에서 `경유정류소목록조회` 로 위에서 확인한 버스 ID 입력 

```python
routeId = 234000878
routeId = str(routeId)
url_busstop = ('https://apis.data.go.kr/6410000/busrouteservice/v2/getBusRouteStationListv2?serviceKey=' + key 
                + '&routeId=' + routeId + '&format=xml')
url_busstop
```

찾기를 원하는 버스의 `<stationId>206000261</stationId>` 와 `<stationSeq>13</stationSeq>`를 확인하세요

만약 버스정류장의 `<mobileNo> 07152</mobileNo>`를 알고 있다면 아래 코드를 통해 stationId와 stationSeq를 찾을 수 있습니다. 

```python
import subprocess
import xml.etree.ElementTree as ET


# curl 명령 실행
result = subprocess.run(["curl", "-X", "GET", url_busstop, "-H", "accept: */*"], capture_output=True, text=True)

# 결과 데이터 저장
xml_data = result.stdout

# XML 파싱
root = ET.fromstring(xml_data)

# 원하는 mobileNo 찾기
mobile_no = "29873"  # 이 부분을 찾고자 하는 mobileNo로 설정

for station in root.findall('.//busRouteStationList'):
    # mobileNo 요소 찾기
    mobile_no_element = station.find('mobileNo')
    
    # mobileNo 요소와 해당 요소의 텍스트가 존재하는지 검사
    if mobile_no_element is not None and mobile_no_element.text and mobile_no_element.text.strip() == mobile_no:
        station_id = station.find('stationId').text
        station_seq = station.find('stationSeq').text
        print(f"stationId: {station_id}, stationSeq: {station_seq}")

```
